## Silhouette Score nad skaliranim podacima

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns

scaler = MinMaxScaler()

df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
```

```python
from sklearn_extra.cluster import KMedoids
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import pandas as pd

silhouette_scores = []
range_n_clusters = range(2, 11)

for n_clusters in range_n_clusters:
    pam = KMedoids(n_clusters=n_clusters, random_state=42)
    cluster_labels = pam.fit_predict(df)
    silhouette_avg = silhouette_score(df, cluster_labels)
    silhouette_scores.append(silhouette_avg)
    print(f"Silhouette Score for {n_clusters} clusters: {silhouette_avg}")

plt.figure(figsize=(8, 5))
plt.plot(range_n_clusters, silhouette_scores, marker='o')
plt.xticks(range_n_clusters)
plt.title('Silhouette Scores for Different Numbers of Clusters (PAM)')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.grid(True)
plt.show()
```

## Elbow metoda nad skaliranim podacima

```python
from sklearn_extra.cluster import KMedoids
import matplotlib.pyplot as plt
import pandas as pd

inertia_values = []
range_n_clusters = range(2, 11)

for n_clusters in range_n_clusters:
    pam = KMedoids(n_clusters=n_clusters, random_state=42)
    pam.fit(df)
    inertia_values.append(pam.inertia_)
    print(f"Inertia for {n_clusters} clusters: {pam.inertia_}")

plt.figure(figsize=(8, 5))
plt.plot(range_n_clusters, inertia_values, marker='o')
plt.xticks(range_n_clusters)
plt.title('Elbow Method for Optimal Number of Clusters (PAM)')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.grid(True)
plt.show()
```

## PAM klastering (k=2)

```python
from sklearn_extra.cluster import KMedoids
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
import pandas as pd
import numpy as np

n_clusters = 2
pam = KMedoids(n_clusters=n_clusters, random_state=42)
cluster_labels = pam.fit_predict(df)

# Evaluacija PAM klasteringa
# 1. Silhouette Score
silhouette_avg = silhouette_score(df, cluster_labels)
print(f"Silhouette Score for {n_clusters} clusters: {silhouette_avg}")

# 2. Davies-Bouldin Index
db_index = davies_bouldin_score(df, cluster_labels)
print(f"Davies-Bouldin Index for {n_clusters} clusters: {db_index}")

# 3. Calinski-Harabasz Index
ch_index = calinski_harabasz_score(df, cluster_labels)
print(f"Calinski-Harabasz Index for {n_clusters} clusters: {ch_index}")

# 4. Intra-klaster odstupanje (WCSS)
wcss = 0
for i in range(n_clusters):
    cluster_points = df[cluster_labels == i]
    medoid = cluster_points.mean(axis=0)
    wcss += np.sum(np.linalg.norm(cluster_points - medoid, axis=1)**2)

print(f"Intra-cluster Sum of Squares (WCSS) for {n_clusters} clusters: {wcss}")
```
