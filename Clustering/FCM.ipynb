##Silhouette score

```python
from sklearn.metrics import silhouette_score
import numpy as np
import skfuzzy as fuzz
import matplotlib.pyplot as plt

data_t = df.T  # Transponovanje podataka za FCM

range_n_clusters = range(2, 11)

silhouette_scores = []

for n_clusters in range_n_clusters:
    cntr, u, _, _, _, _, _ = fuzz.cluster.cmeans(
        data_t, n_clusters, 2, error=0.005, maxiter=1000, init=None
    )

    hard_labels = np.argmax(u, axis=0)

    silhouette_avg = silhouette_score(df, hard_labels)
    silhouette_scores.append(silhouette_avg)
    print(f"Silhouette Score for {n_clusters} clusters: {silhouette_avg}")

plt.figure(figsize=(8, 5))
plt.plot(range_n_clusters, silhouette_scores, marker='o', label='Silhouette Score')
plt.title('Silhouette Scores for Different Numbers of Clusters (Fuzzy C-Means)')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.xticks(range_n_clusters)
plt.grid(True)
plt.legend()
plt.show()
```

##Elbow metoda

```python
import matplotlib.pyplot as plt
import numpy as np
import skfuzzy as fuzz

data_t = df.T

wss = []
range_n_clusters = range(2, 11)

print("WSS values for each number of clusters:")
for n_clusters in range_n_clusters:
    cntr, u, _, _, _, _, _ = fuzz.cluster.cmeans(
        data_t, n_clusters, 2, error=0.005, maxiter=1000, init=None
    )
    # WSS: Zbir kvadratnih udaljenosti ponderisanih fuzzy pripadnostima
    wss_value = np.sum(np.min(u**2, axis=0))
    wss.append(wss_value)
    print(f"Number of clusters {n_clusters}: WSS = {wss_value}")

plt.figure(figsize=(8, 5))
plt.plot(range_n_clusters, wss, marker='o')
plt.title('Elbow Method for Fuzzy C-Means')
plt.xlabel('Number of Clusters')
plt.ylabel('Within-Cluster Sum of Squares (WSS)')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import skfuzzy as fuzz
import matplotlib.pyplot as plt

# Transponovanje podataka za FCM
data_t = df.T

# Parametri
n_clusters = 3  # Broj klastera (promenjeno sa 2 na 5)
m_values = [1.5, 2, 3]  # Različite vrednosti m
error = 0.005
maxiter = 1000

# Poređenje za različite vrednosti m
for m in m_values:
    # Primena Fuzzy C-Means
    cntr, u, _, _, _, _, _ = fuzz.cluster.cmeans(
        data_t, n_clusters, m=m, error=error, maxiter=maxiter, init=None
    )

    # Izračunavanje Fuzzy Partition Coefficient (FPC)
    fpc = np.sum(u**2) / u.shape[1]
    print(f"Fuzzy Partition Coefficient (FPC) for m={m}: {fpc}")

    # Prikaz fuzzy membership-a za svaki klaster
    plt.figure(figsize=(10, 6))

    # Prikazivanje svih klastera na jednom scatter plotu
    for i in range(n_clusters):
        plt.scatter(range(data_t.shape[1]), u[i, :], label=f'Cluster {i+1}', alpha=0.6)

    # Dodavanje naslova i oznaka
    plt.title(f'Membership Distribution for k={n_clusters} Clusters (Fuzzy C-Means, m={m})')
    plt.xlabel('Samples')
    plt.ylabel('Membership Value')
    plt.legend()
    plt.grid(True)
    plt.show()
```
